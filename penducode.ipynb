```python
import random

def choix_mots(liste): #fonction qui renvoie le mot a trouver
  a = random.randint(0, len(liste) - 1) #choix d'un entier au hasard, utilisé pour piocher au hasard dans la liste
  mot = liste[a] #CHOIX DU MOT
  print("Le mot contient", len(mot), "lettres. Bonne chance!") # longeur du mot
  return mot

#---BOUCLE PRINCIPALE---

def jeu_du_pendu():
  mot_a_deviner = choix_mots(["musique", "cheval", "fleur"])
  compteur_mauvaises_lettres = 0 # Initialisation du score pour les mauvaises tentatives
  lettres_devinees = [] # Pour stocker les lettres que le joueur a déjà proposées
  mot_actuel_affiche = ["_" for _ in mot_a_deviner] # Représentation du mot avec des underscores

  # Replaced ' '.join(affichage_mot) with a loop for easier understanding
  mot_initial_display = ""
  for char_to_show in mot_actuel_affiche:
    mot_initial_display = mot_initial_display + char_to_show + " "
  print("Mot à trouver: ", mot_initial_display.strip())

  while compteur_mauvaises_lettres < 7 and "_" in mot_actuel_affiche: # Boucle principale du jeu
    l = str(input("Quelle lettre choisissez-vous? ")).lower() # Convertir en minuscule pour uniformité

    # Replaced len(l) != 1 or not l.isalpha() with a more basic check
    is_single_letter = True
    if len(l) != 1:
      is_single_letter = False
    else:
      char_code = ord(l)
      # Check if character's ASCII value falls within 'a' to 'z'
      if not (ord('a') <= char_code <= ord('z')):
        is_single_letter = False

    if not is_single_letter:
      print("Veuillez entrer une seule lettre de l'alphabet.")
      continue

    if l in lettres_devinees:
      print("Vous avez déjà proposé cette lettre.")
      continue

    lettres_devinees.append(l) # Ajouter la lettre à la liste des lettres déjà devinées

    if l in mot_a_deviner:
      print("La lettre est dans le mot!")
      # Replaced enumerate with a basic for loop for easier understanding
      for i in range(len(mot_a_deviner)):
        if mot_a_deviner[i] == l:
          mot_actuel_affiche[i] = l # Révéler la lettre dans le mot affiché

      # Replaced ' '.join(mot_actuel_affiche) with a loop for easier understanding
      current_word_display = ""
      for char_to_show in mot_actuel_affiche:
        current_word_display = current_word_display + char_to_show + " "
      print("Mot actuel: ", current_word_display.strip())
    else:
      print("Raté!")
      compteur_mauvaises_lettres += 1 # Incrémenter le compteur de mauvaises lettres
      print(f"Il vous reste {7 - compteur_mauvaises_lettres} chances.")

  if "_" not in mot_actuel_affiche:
    print("Félicitations! Vous avez trouvé le mot: ", mot_a_deviner)
  else:
    print("Vous avez perdu. Le mot était: ", mot_a_deviner)

  volonte_rejouer = input("Voulez-vous rejouer? (oui/non) ").lower()
  return volonte_rejouer == 'oui'

# Boucle pour rejouer
jouer_encore = True
while jouer_encore:
  jouer_encore = jeu_du_pendu()
print("Merci d'avoir joué!")
```
